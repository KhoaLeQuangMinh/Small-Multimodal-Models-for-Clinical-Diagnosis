```python
import os
from huggingface_hub import hf_hub_download

# Define the exact target repository
repo = "mauro-nievoff/MultiCaRe_Dataset"

# List of critical missing tabular metadata documents
missing_files = [
    "captions_and_labels.csv",
    "cases.parquet",
    "abstracts.parquet",
    "case_images.parquet",
    "data_dictionary.csv"
]

print("Starting targeted sync for missing metadata files...")

for file in missing_files:
    try:
        print(f"🔄 Downloading: {file}...")
        hf_hub_download(
            repo_id=repo,
            filename=file,
            repo_type="dataset"
        )
        print(f"✅ Successfully cached {file}")
    except Exception as e:
        print(f"❌ Failed to fetch {file}: {e}")

print("\n🎉 Sync Complete! Re-check your Finder folder now.")
```
